# Conformal Prediction Demo
Quantificazione incertezza distribution-free.

```python
from src.evaluation.conformal import ConformalConfig, create_conformal_predictor
import numpy as np
y_true = np.random.rand(100)
y_pred = y_true + np.random.normal(0, 0.1, 100)
cp = create_conformal_predictor(ConformalConfig(alpha=0.1, method='enbpi'))
cp.fit(y_true, y_pred)
low, high = cp.predict_interval(y_pred)
print(f"Coverage: {np.mean((y_true>=low) & (y_true<=high)):.2%}")
```
